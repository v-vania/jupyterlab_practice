```python
def compute_mean(value_list):
    mean = sum(value_list)/len(value_list)
    return mean

compute_mean([2, 4, 6, 8])
```

The above code cell calculates the mean of the supplied list of values by dividing the sum of the values by the number of values.
$$
\begin{flalign}
\bar{x} = \frac{x_1 + x_2 + \dots + x_n}{n} = \frac{1}{n}\bigg(\sum_{i=1}^{n}x_i \bigg)&&
\end{flalign}
$$
